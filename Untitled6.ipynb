## Steps 3 & 4: Data Preprocessing and Exploratory Data Analysis (EDA)

We will preprocess the textual fields and analyze distribution, suspicious keywords, and features associated with fraudulence.

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Ensure all necessary NLTK packages are downloaded
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

# 1. Missing Value and Duplicate Handling
df = df.drop_duplicates().copy()  # Use copy to avoid setting with copy warnings
df.fillna('', inplace=True)

# 2. Text Preprocessing
stop_words = set(stopwords.words('english'))

def clean_text(text):
    if not text:
        return ""
    text = str(text).lower()
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)  # remove special chars
    tokens = word_tokenize(text)
    cleaned_tokens = [w for w in tokens if w not in stop_words]
    return " ".join(cleaned_tokens)

# Create combined text column for TF-IDF training
df['cleaned_profile'] = df['company_profile'].apply(clean_text)
# Combine text features to simulate realistic job profile information
df['text_features'] = df['title'] + " " + df['cleaned_profile'] + " " + df['description'].apply(clean_text) + " " + df['requirements'].apply(clean_text)

print("Preprocessing Complete!")

# 3. Exploratory Data Analysis Plots
plt.figure(figsize=(15, 5))

# Chart 1: Distribution of Labels
plt.subplot(1, 2, 1)
sns.countplot(data=df, x='fraudulent', palette='Set1')
plt.title('Genuine (0) vs Fraudulent (1) Postings')
plt.xlabel('Status')
plt.ylabel('Count')

# Chart 2: Fraud distribution by Employment Type
plt.subplot(1, 2, 2)
sns.countplot(data=df, x='employment_type', hue='fraudulent', palette='Set2')
plt.title('Fraud Distribution by Employment Type')
plt.xlabel('Employment Type')
plt.ylabel('Count')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## Steps 5 & 6: Machine Learning Algorithms and Model Evaluation

We convert text into vector representations using TF-IDF and train Logistic Regression, Naive Bayes, Random Forest, and Support Vector Machine models. We evaluate them on multiple classification metrics with a strong focus on Recall.

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix
import joblib

# Split features and target labels
X = df['text_features']
y = df['fraudulent']

# Train-test split (reproducible with random_state)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# TF-IDF Feature Extraction
tfidf = TfidfVectorizer(max_features=2500, min_df=2)
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

# Initialize models
models = {
    'Logistic Regression': LogisticRegression(random_state=42),
    'Naive Bayes': MultinomialNB(),
    'Random Forest': RandomForestClassifier(random_state=42, n_estimators=100),
    'Support Vector Machine': SVC(random_state=42, probability=True)
}

# Store metrics to compare results
metrics_summary = {}

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.ravel()

for idx, (name, model) in enumerate(models.items()):
    # Train
    model.fit(X_train_tfidf, y_train)
    # Predict
    preds = model.predict(X_test_tfidf)
    probs = model.predict_proba(X_test_tfidf)[:, 1]

    # Evaluate metrics
    acc = accuracy_score(y_test, preds)
    prec = precision_score(y_test, preds, zero_division=0)
    rec = recall_score(y_test, preds, zero_division=0)
    f1 = f1_score(y_test, preds, zero_division=0)
    auc = roc_auc_score(y_test, probs)

    metrics_summary[name] = {
        'Accuracy': acc,
        'Precision': prec,
        'Recall': rec,
        'F1-Score': f1,
        'ROC-AUC': auc
    }

    # Confusion Matrix
    cm = confusion_matrix(y_test, preds)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[idx],
                xticklabels=['Genuine', 'Fraudulent'], yticklabels=['Genuine', 'Fraudulent'])
    axes[idx].set_title(f"{name}\nRecall: {rec:.2f} | F1: {f1:.2f}")
    axes[idx].set_xlabel('Predicted')
    axes[idx].set_ylabel('Actual')

plt.tight_layout()
plt.show()

# Print metrics comparison table
df_metrics = pd.DataFrame(metrics_summary).T
print("=== MODEL PERFORMANCE COMPARISON ===")
display(df_metrics)

# Save the best model (Random Forest in this mock/real example) and the TF-IDF Vectorizer for application use
joblib.dump(models['Random Forest'], 'best_fake_job_detector.pkl')
joblib.dump(tfidf, 'tfidf_vectorizer.pkl')
print("Models and vectorizers saved successfully!")
```

## Steps 7, 8 & 9: Real-time Prediction System, Explainable AI, and Advanced Analytics

We design a diagnostic function that calculates an advanced risk score, highlights suspicious attributes, provides prediction confidence, and generates a structured report.

```python
import re

def analyze_job_posting(title, company_name, description, salary, application_url, company_website, contact_email):
    # 1. Traditional Verification Check / Heuristics
    suspicious_flags = []
    verified_signals = []
    heuristics_score = 0

    # Check for unverified email domains (e.g., free public domains like gmail/yahoo for a supposed firm)
    free_domains = ['gmail.com', 'yahoo.com', 'hotmail.com', 'outlook.com', 'instant-income.xyz', 'easyjobs.info']
    email_domain = contact_email.split('@')[-1] if '@' in contact_email else ""

    if not contact_email:
        suspicious_flags.append("Missing contact email address.")
        heuristics_score += 20
    elif email_domain in free_domains:
        suspicious_flags.append(f"Uses a free or suspicious email domain: @{email_domain}")
        heuristics_score += 25
    else:
        verified_signals.append("Uses a custom official business email domain.")

    # Check for company website presence
    if not company_website or company_website.strip() == "":
        suspicious_flags.append("No official company website URL provided.")
        heuristics_score += 20
    else:
        verified_signals.append(f"Company website listed: {company_website}")

    # Check for suspicious keywords in description
    suspicious_keywords = ['guaranteed payout', 'pay registration fee', 'wire money', 'deposit first', 'no experience needed start making', 'instant money', 'work from home helper']
    desc_lower = description.lower()
    found_keywords = [kw for kw in suspicious_keywords if kw in desc_lower]
    if found_keywords:
        suspicious_flags.append(f"Description contains urgent financial/get-rich-quick buzzwords: {found_keywords}")
        heuristics_score += 30

    # 2. ML Model Prediction Integration
    model_input = f"{title} {description} {salary}"
    cleaned_input = clean_text(model_input)
    vec_input = tfidf.transform([cleaned_input])

    # Using our trained Random Forest classifier
    rf_model = models['Random Forest']
    pred_prob = rf_model.predict_proba(vec_input)[0][1]

    # Combine Heuristics (40%) and ML Probability (60%) for a balanced, robust Risk Score
    ml_contribution = pred_prob * 100
    final_risk_score = min(100, (heuristics_score * 0.4) + (ml_contribution * 0.6))

    # 3. Risk Classification
    if final_risk_score < 35:
        status = "Genuine"
        color = "Green"
    elif final_risk_score < 70:
        status = "Suspicious"
        color = "Orange"
    else:
        status = "High Risk"
        color = "Red"

    # Generate Detailed Report
    print("===========================================================")
    print("       JOB ADVERTISEMENT RISK ASSESSMENT REPORT            ")
    print("===========================================================")
    print(f"Job Title: {title}")
    print(f"Company Name: {company_name}")
    print(f"Overall Risk Score: {final_risk_score:.1f} / 100 ({status})")
    print(f"ML Prediction Confidence: {pred_prob*100:.1f}%")
    print("-----------------------------------------------------------")
    print("Verified Positive Signals:")
    for signal in (verified_signals if verified_signals else ["None"]):
        print(f" [+] {signal}")
    print("-----------------------------------------------------------")
    print("Suspicious Flags detected:")
    for flag in (suspicious_flags if suspicious_flags else ["None (Good profile structure)"]):
        print(f" [-] {flag}")
    print("===========================================================\n")

    return {
        'status': status,
        'risk_score': final_risk_score,
        'confidence': pred_prob * 100,
        'suspicious_flags': suspicious_flags,
        'verified_signals': verified_signals
    }

# Quick Interactive Test
test_profile = analyze_job_posting(
    title="Work from Home Assistant Needed",
    company_name="Instant Profit LLC",
    description="We need urgent helpers to work from home typing words. Anyone can apply. No experience required. Start immediately. Must pay standard registration fee for software materials.",
    salary="$5,000 per week guaranteed!",
    application_url="http://instant-income.xyz/join",
    company_website="",
    contact_email="urgentworknow11@gmail.com"
)
```
